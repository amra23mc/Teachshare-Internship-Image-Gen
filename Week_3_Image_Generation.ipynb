Imports


```python
from diffusers import AutoPipelineForText2Image
import torch
from PIL import Image
```

Input Function:

```python
prompt = input("Enter prompt: ")
```

Generation function

```python
def generate(prompt):
    model = AutoPipelineForText2Image.from_pretrained("segmind/Segmind-Vega",torch_dtype=torch.float16, use_safetensors=True)

    model = model.to("cuda" if torch.cuda.is_available() else "cpu")

    image = model(prompt=prompt,  width=1024,height=1024,guidance_scale=7.5,).images[0]

    display(image)

    return image
```
